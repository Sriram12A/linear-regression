```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import linear_model
```

```python
df=pd.read_csv('Book1.csv')
```

```python
df.head()
```

```python
df.fillna(0, inplace=True)
```

```python
reg1= linear_model.LinearRegression()
reg1.fit(df[['area','bedrooms','age']],df.price)
```

```python
plt.scatter(df.area,df.price,color='red')
```

```python
reg1.coef_
```

```python
d=pd.read_csv('test set.csv')
```

```python
reg1.predict(d[['area', 'bedrooms', 'age']])
```

```python
d
```

```python
d['price']=reg1.predict(d[['area','bedrooms','age']])
```

```python
d
```

```python
d.to_csv('test set.csv',index=False)
```

```python
plt.scatter(df[['area']],df['price'],color='red')
plt.plot(df[['area','bedrooms','age']],reg1.predict(df[['area', 'bedrooms', 'age']]),color='red')
```

```python
p=pd.read_csv('hiring.csv')
```

```python
reg=linear_model.LinearRegression()
reg.fit(p[['experience', 'test_score(out of 10)', 'interview_score(out of 10)']], p['salary($)'])
```
